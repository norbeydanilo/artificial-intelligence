Este notebook presenta ejemplos prácticos en español de:
- Zero-Shot
- One-Shot
- Few-Shot

Todos usan la librería `openai` (v1.x) con `client.chat.completions.create()`.

````python
# Carga las variables del archivo .env y crea el cliente
from dotenv import load_dotenv
import os
from openai import OpenAI

load_dotenv()  # Lee el archivo .env de la carpeta actual

client = OpenAI(api_key=os.environ.get("OPENAI_API_KEY"))

print("Cliente configurado correctamente.")
````

---

## 🔹 Zero-Shot

**Concepto:** el modelo recibe solo la tarea, sin ningún ejemplo previo. Funciona bien cuando la tarea es clara y el modelo ya tiene suficiente conocimiento sobre ella.

````python
prompt = "Clasifica el sentimiento del siguiente texto: 'No me gustó el servicio, fue muy lento y confuso.'"

response = client.chat.completions.create(
    model="gpt-5.4-mini",
    messages=[{"role": "user", "content": prompt}]
)

print(response.choices[0].message.content)
````

---

## 🔸 One-Shot

**Concepto:** se proporciona **un solo ejemplo** del formato de entrada y salida esperado. Útil cuando la tarea requiere un patrón de respuesta específico.

````python
prompt = (
    "Ejemplo:\n"
    "Texto: 'Me encantó la comida, todo estuvo delicioso.'\n"
    "Sentimiento: Positivo\n\n"
    "Texto: 'No me gustó el servicio, fue muy lento y confuso.'\n"
    "Sentimiento:"
)

response = client.chat.completions.create(
    model="gpt-5.4-mini",
    messages=[{"role": "user", "content": prompt}]
)

print(response.choices[0].message.content)
````

---

## 🔹 Few-Shot

**Concepto:** se dan **varios ejemplos** para que el modelo identifique el patrón antes de responder. Mejora la consistencia y la precisión cuando hay múltiples categorías o formatos posibles.

````python
prompt = (
    "Texto: 'Me encantó la comida, todo estuvo delicioso.'\n"
    "Sentimiento: Positivo\n\n"
    "Texto: 'El lugar estaba sucio y el mesero fue grosero.'\n"
    "Sentimiento: Negativo\n\n"
    "Texto: 'La experiencia fue normal, nada fuera de lo común.'\n"
    "Sentimiento: Neutral\n\n"
    "Texto: 'No me gustó el servicio, fue muy lento y confuso.'\n"
    "Sentimiento:"
)

response = client.chat.completions.create(
    model="gpt-5.4-mini",
    messages=[{"role": "user", "content": prompt}]
)

print(response.choices[0].message.content)
````
